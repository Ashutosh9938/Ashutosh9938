````python
import json
import re

def extract_order(message):
    messages = [
        {
            "role": "system",
            "content": (
                "Extract restaurant orders from user messages. "
                "Return only a valid JSON object, without Markdown. "
                "Required keys: tableNumber, customerCount, items. "
                "Each item must have name and quantity. "
                "Do not invent missing values. "
                "Use null for unknown tableNumber or customerCount."
            )
        },
        {
            "role": "user",
            "content": message
        }
    ]

    result = model(
        text=messages,
        max_new_tokens=256
    )

    output = result[0]["generated_text"][-1]["content"].strip()

    # Remove Markdown code fences if the model adds them.
    output = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        output
    ).strip()

    order = json.loads(output)

    assert isinstance(order["items"], list)

    for item in order["items"]:
        assert isinstance(item["name"], str)
        assert isinstance(item["quantity"], int)
        assert item["quantity"] > 0

    return order
````

````python

import json
import re
from typing import Optional

from pydantic import BaseModel, Field, ConfigDict, ValidationError


# ============================================
# 1. DEFINE ORDER SCHEMA
# ============================================

class OrderItem(BaseModel):
    model_config = ConfigDict(strict=True, extra="forbid")

    name: str = Field(min_length=1)
    quantity: int = Field(gt=0)


class RestaurantOrder(BaseModel):
    model_config = ConfigDict(strict=True, extra="forbid")

    tableNumber: Optional[int]
    customerCount: Optional[int] = Field(gt=0)
    items: list[OrderItem]


# ============================================
# 2. SYSTEM PROMPT
# ============================================

SYSTEM_PROMPT = """
You are a restaurant order extraction assistant.

Your job is to convert natural language restaurant
orders into structured JSON.

STRICT RULES:

1. Return ONLY valid JSON without Markdown.
2. Never invent missing information.
3. If the table number is not explicitly mentioned,
   return null.
4. If the customer count is not explicitly mentioned,
   return null.
5. Table numbers and quantities must be integers.
6. Never infer customer count from item quantities.
7. Extract only food and drinks requested by the user.
8. Never calculate prices.
9. Never add items that were not requested.
10. Preserve item quantities correctly.
11. If a field is unknown, use null rather than guessing.

REQUIRED OUTPUT FORMAT:

{
  "tableNumber": null,
  "customerCount": null,
  "items": [
    {
      "name": "example item",
      "quantity": 1
    }
  ]
}

EXAMPLE 1:

Input:
Table 2 needs three pizzas and two coffees.

Output:
{
  "tableNumber": 2,
  "customerCount": null,
  "items": [
    {"name": "pizza", "quantity": 3},
    {"name": "coffee", "quantity": 2}
  ]
}

EXAMPLE 2:

Input:
Add two burgers and one Coke.

Output:
{
  "tableNumber": null,
  "customerCount": null,
  "items": [
    {"name": "burger", "quantity": 2},
    {"name": "Coke", "quantity": 1}
  ]
}

EXAMPLE 3:

Input:
Four people at table 7 ordered four momos.

Output:
{
  "tableNumber": 7,
  "customerCount": 4,
  "items": [
    {"name": "momo", "quantity": 4}
  ]
}

EXAMPLE 4:

Input:
Table 5 has three customers. Two want chicken
momo and the other wants fried rice.

Output:
{
  "tableNumber": 5,
  "customerCount": 3,
  "items": [
    {"name": "chicken momo", "quantity": 2},
    {"name": "fried rice", "quantity": 1}
  ]
}
"""


# ============================================
# 3. EXTRACT ORDER FUNCTION
# ============================================

def extract_order(message: str) -> dict:

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": message
        }
    ]

    # Generate the model response
    response = model(
        text=messages,
        max_new_tokens=256,
        generate_kwargs={
            "do_sample": False
        }
    )

    # Extract generated text
    output = response[0]["generated_text"][-1]["content"]

    # Remove Markdown code blocks if present
    output = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        output.strip()
    ).strip()

    # Parse JSON
    data = json.loads(output)

    # Validate against Pydantic schema
    order = RestaurantOrder.model_validate(data)

    return order.model_dump()


# ============================================
# 4. TEST CASES
# ============================================

tests = [
    "Table 2 needs three pizzas and two coffees.",

    "We have four people on table 7. "
    "Two want chicken momo and the others want fried rice.",

    "Add one Coke and two burgers to my order.",

    "Table 5 has 3 customers. "
    "They want 2 chicken momos, 1 burger and 3 Cokes.",

    "Give me two coffees and one sandwich.",

    "There are six customers at table 9. "
    "Three ordered chicken burgers and the remaining "
    "three ordered veg momo."
]


# ============================================
# 5. RUN TESTS
# ============================================

for message in tests:

    print("\n" + "=" * 60)
    print("INPUT:", message)

    try:
        result = extract_order(message)

        print("OUTPUT:")
        print(json.dumps(result, indent=2))

    except (ValidationError, ValueError, KeyError, TypeError) as e:
        print("VALIDATION FAILED:", e)

print("\nTesting completed.")
````

````python

import re
from decimal import Decimal


def normalize_name(name: str) -> str:
    name = name.lower().strip()
    name = re.sub(r"[^a-z0-9\s]", "", name)
    return " ".join(name.split())


def match_menu_item(item_name: str, menu: list) -> dict:
    requested = normalize_name(item_name)

    # Handle simple plural forms
    variants = {requested}

    if requested.endswith("s"):
        variants.add(requested[:-1])

    matches = []

    for menu_item in menu:
        menu_name = normalize_name(menu_item["name"])

        if menu_name in variants:
            matches.append(menu_item)

    if len(matches) == 1:
        item = matches[0]

        if not item["available"]:
            return {
                "status": "unavailable",
                "requestedName": item_name,
                "menuItemId": item["id"]
            }

        return {
            "status": "matched",
            "requestedName": item_name,
            "menuItemId": item["id"],
            "name": item["name"],
            "unitPrice": item["price"]
        }

    # Find possible options for an ambiguous name
    possible = [
        item
        for item in menu
        if requested in normalize_name(item["name"])
    ]

    if possible:
        return {
            "status": "ambiguous",
            "requestedName": item_name,
            "options": [
                {
                    "id": item["id"],
                    "name": item["name"],
                    "available": item["available"]
                }
                for item in possible
            ]
        }

    return {
        "status": "not_found",
        "requestedName": item_name
    }


def prepare_order_draft(message: str) -> dict:
    extracted = extract_order(message)

    matched_items = []
    issues = []
    subtotal = Decimal("0")

    for requested_item in extracted["items"]:
        match = match_menu_item(
            requested_item["name"],
            restaurant_menu
        )

        if match["status"] != "matched":
            issues.append({
                **match,
                "quantity": requested_item["quantity"]
            })
            continue

        quantity = requested_item["quantity"]
        unit_price = Decimal(str(match["unitPrice"]))
        line_total = unit_price * quantity

        matched_items.append({
            "menuItemId": match["menuItemId"],
            "name": match["name"],
            "quantity": quantity,
            "unitPrice": str(unit_price),
            "lineTotal": str(line_total)
        })

        subtotal += line_total

    return {
        "tableNumber": extracted["tableNumber"],
        "customerCount": extracted["customerCount"],
        "items": matched_items,
        "issues": issues,
        "subtotal": str(subtotal),
        "readyForConfirmation": (
            len(issues) == 0 and len(matched_items) > 0
        )
    }
````

````python

import json
import re
from decimal import Decimal
from pydantic import BaseModel, ConfigDict, Field


# Our temporary in-memory conversation state
order_session = None


def get_subtotal(items):
    return sum(
        (
            Decimal(item["unitPrice"]) * item["quantity"]
            for item in items
        ),
        Decimal("0")
    )


def show_order(session):
    print("\n--- CURRENT ORDER ---")
    print("Table:", session["tableNumber"])
    print("Status:", session["status"])

    for item in session["items"]:
        print(f'{item["quantity"]} x {item["name"]}')

    print("Subtotal: Rs.", get_subtotal(session["items"]))


def next_question(session):
    if not session["pendingItems"]:
        session["status"] = "awaiting_confirmation"

        return (
            f'Order subtotal: Rs. {get_subtotal(session["items"])}. '
            "Would you like to confirm?"
        )

    issue = session["pendingItems"][0]
    session["status"] = "awaiting_clarification"

    if issue["status"] == "ambiguous":
        options = ", ".join(
            item["name"]
            for item in issue["options"]
            if item["available"]
        )

        return (
            f'You requested {issue["quantity"]} '
            f'{issue["requestedName"]}. '
            f'Which would you like? Options: {options}'
        )

    if issue["status"] == "unavailable":
        return (
            f'{issue["requestedName"]} is unavailable. '
            "Please ask staff to remove or replace it."
        )

    return (
        f'{issue["requestedName"]} was not found on the menu. '
        "Please ask staff to remove or replace it."
    )


def start_order(message):
    global order_session

    draft = prepare_order_draft(message)

    order_session = {
        "tableNumber": draft["tableNumber"],
        "customerCount": draft["customerCount"],
        "items": draft["items"].copy(),
        "pendingItems": draft["issues"].copy(),
        "status": "draft"
    }

    show_order(order_session)
    print("\nASSISTANT:", next_question(order_session))

    return order_session
````

````python

class ClarificationSelection(BaseModel):
    model_config = ConfigDict(strict=True, extra="forbid")

    menuItemId: int
    quantity: int = Field(gt=0)


class ClarificationResult(BaseModel):
    model_config = ConfigDict(strict=True, extra="forbid")

    selections: list[ClarificationSelection]


def understand_clarification(message, pending_item):
    options = [
        {
            "menuItemId": item["id"],
            "name": item["name"]
        }
        for item in pending_item["options"]
        if item["available"]
    ]

    prompt = f"""
You resolve ambiguous restaurant menu choices.

The customer previously requested:
{pending_item["quantity"]} x {pending_item["requestedName"]}

Available options:
{json.dumps(options)}

Customer reply:
{message}

RULES:
1. Return only valid JSON.
2. Use only menuItemId values from available options.
3. Never invent items, quantities, or IDs.
4. Interpret quantities from the customer's reply.
5. If the reply is unclear, return an empty selections list.
6. If quantities are missing or uncertain, return
   an empty selections list.
7. Do not guess.

Example:
Options:
[
  {{"menuItemId": 1, "name": "Chicken Burger"}},
  {{"menuItemId": 2, "name": "Veg Burger"}}
]

Reply: One chicken and one veg.

Output:
{{
  "selections": [
    {{"menuItemId": 1, "quantity": 1}},
    {{"menuItemId": 2, "quantity": 1}}
  ]
}}

Now resolve the actual customer reply.
"""

    response = model(
        text=[
            {
                "role": "system",
                "content": (
                    "You are a precise menu clarification "
                    "parser. Return only JSON."
                )
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_new_tokens=256,
        generate_kwargs={"do_sample": False}
    )

    output = response[0]["generated_text"][-1]["content"]

    output = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        output.strip()
    ).strip()

    parsed = ClarificationResult.model_validate(
        json.loads(output)
    )

    return parsed.model_dump()
````

````python

def reply_to_order(message):
    global order_session

    if order_session is None:
        print("No active order. Call start_order() first.")
        return

    if order_session["status"] != "awaiting_clarification":
        print("Order is not awaiting clarification.")
        return

    pending = order_session["pendingItems"][0]

    if pending["status"] != "ambiguous":
        print("This issue cannot be resolved by selection yet.")
        print(next_question(order_session))
        return

    try:
        result = understand_clarification(message, pending)
        selections = result["selections"]

        if not selections:
            print("ASSISTANT: I couldn't understand your choices.")
            print(next_question(order_session))
            return

        # Restrict choices to the options we supplied
        allowed_ids = {
            item["id"]
            for item in pending["options"]
            if item["available"]
        }

        selected_ids = [
            item["menuItemId"]
            for item in selections
        ]

        if (
            len(selected_ids) != len(set(selected_ids))
            or not set(selected_ids).issubset(allowed_ids)
        ):
            print("Invalid or duplicate menu item selection.")
            return

        # The quantity cannot silently change
        total_quantity = sum(
            item["quantity"] for item in selections
        )

        if total_quantity != pending["quantity"]:
            print(
                "ASSISTANT: Please clarify all",
                pending["quantity"],
                "items."
            )
            return

        # Prepare changes before updating the session
        new_items = []

        for selection in selections:
            menu_item = next(
                (
                    item for item in restaurant_menu
                    if item["id"] == selection["menuItemId"]
                ),
                None
            )

            if menu_item is None or not menu_item["available"]:
                print("Selected item is unavailable.")
                return

            unit_price = Decimal(str(menu_item["price"]))
            quantity = selection["quantity"]

            new_items.append({
                "menuItemId": menu_item["id"],
                "name": menu_item["name"],
                "quantity": quantity,
                "unitPrice": str(unit_price),
                "lineTotal": str(unit_price * quantity)
            })

        # Apply the validated changes
        order_session["items"].extend(new_items)
        order_session["pendingItems"].pop(0)

        show_order(order_session)

        print("\nASSISTANT:", next_question(order_session))

    except Exception as error:
        print("Unable to process clarification:", error)
        print("Please try describing your choices again.")
````

````python

import re

def confirm_order(message):
    global order_session

    if order_session is None:
        return "No active order."

    if order_session["status"] != "awaiting_confirmation":
        return "Your order is not ready for confirmation."

    # Normalize message
    response = message.lower().strip()

    # Remove punctuation and normalize spaces
    response = re.sub(r"[^\w\s]", "", response)
    response = " ".join(response.split())

    confirmations = {
        "yes",
        "confirm",
        "confirm order",
        "yes confirm",
        "yes confirm my order",
        "place order",
        "please confirm my order",
        "go ahead",
        "yes please"
    }

    cancellations = {
        "no",
        "cancel",
        "cancel order",
        "no cancel",
        "dont place it",
        "please cancel my order"
    }

    if response in confirmations:
        order_session["status"] = "confirmed"

        return (
            f'Order confirmed for table {order_session["tableNumber"]}. '
            f'Subtotal: Rs. {get_subtotal(order_session["items"])}'
        )

    if response in cancellations:
        order_session["status"] = "cancelled"
        return "Order draft cancelled."

    return "Please reply 'confirm' or 'cancel'."
````

````python

def chat(message):
    global order_session

    print(f"\nUSER: {message}")

    # No existing order, or previous order is finished
    if (
        order_session is None
        or order_session["status"] in ["confirmed", "cancelled"]
    ):
        start_order(message)
        return

    # Waiting for a menu clarification
    if order_session["status"] == "awaiting_clarification":
        reply_to_order(message)
        return

    # Waiting for customer confirmation
    if order_session["status"] == "awaiting_confirmation":
        response = confirm_order(message)
        print(f"\nASSISTANT: {response}")
        return

    print("\nASSISTANT: Unable to process the current order.")
````

````python

from uuid import uuid4
from decimal import Decimal

sessions = {}


def create_session(restaurant_id, table_number):
    if restaurant_id not in restaurants:
        raise ValueError("Restaurant does not exist")

    if not isinstance(table_number, int) or table_number <= 0:
        raise ValueError("Invalid table number")

    session_id = str(uuid4())

    sessions[session_id] = {
        "sessionId": session_id,
        "restaurantId": restaurant_id,
        "tableNumber": table_number,
        "customerCount": None,
        "items": [],
        "pendingItems": [],
        "status": "new"
    }

    return session_id


def calculate_subtotal(session):
    return sum(
        (
            Decimal(item["unitPrice"]) * item["quantity"]
            for item in session["items"]
        ),
        Decimal("0")
    )


def get_session(session_id):
    if session_id not in sessions:
        raise ValueError("Invalid session")

    return sessions[session_id]


def show_session(session_id):
    session = get_session(session_id)

    print("\nRESTAURANT:", restaurants[
        session["restaurantId"]
    ]["name"])

    print("TABLE:", session["tableNumber"])
    print("STATUS:", session["status"])

    for item in session["items"]:
        print(
            f'{item["quantity"]} x {item["name"]} '
            f'@ Rs. {item["unitPrice"]}'
        )

    print("SUBTOTAL: Rs.", calculate_subtotal(session))
````

````python

def chat_multi(session_id, message):
    session = get_session(session_id)

    restaurant = restaurants[session["restaurantId"]]
    menu = restaurant["menu"]

    print(f"\nUSER: {message}")

    # CASE 1: New order
    if session["status"] == "new":
        extracted = extract_order(message)

        # Trusted session controls table identity
        session["customerCount"] = extracted["customerCount"]

        # Reject conflicting table references
        if (
            extracted["tableNumber"] is not None
            and extracted["tableNumber"] != session["tableNumber"]
        ):
            print("ASSISTANT: This session belongs to a different table.")
            return

        for requested in extracted["items"]:
            match = match_menu_item(requested["name"], menu)

            if match["status"] == "matched":
                quantity = requested["quantity"]
                price = Decimal(str(match["unitPrice"]))

                session["items"].append({
                    "menuItemId": match["menuItemId"],
                    "name": match["name"],
                    "quantity": quantity,
                    "unitPrice": str(price),
                    "lineTotal": str(price * quantity)
                })
            else:
                session["pendingItems"].append({
                    **match,
                    "quantity": requested["quantity"]
                })

    # CASE 2: Resolve ambiguous menu item
    elif session["status"] == "awaiting_clarification":
        pending = session["pendingItems"][0]

        if pending["status"] != "ambiguous":
            print(
                "ASSISTANT: This item is unavailable or "
                "not on the menu. Please contact staff."
            )
            return

        result = understand_clarification(message, pending)
        selections = result["selections"]

        valid_options = {
            item["id"]: item
            for item in pending["options"]
            if item["available"]
        }

        ids = [s["menuItemId"] for s in selections]

        if (
            not selections
            or len(ids) != len(set(ids))
            or not set(ids).issubset(valid_options)
            or sum(s["quantity"] for s in selections)
                != pending["quantity"]
        ):
            print("ASSISTANT: Please clarify your selections.")
            return

        # Validate against this restaurant's current menu
        new_items = []

        for selection in selections:
            item = next(
                (
                    x for x in menu
                    if x["id"] == selection["menuItemId"]
                    and x["available"]
                ),
                None
            )

            if item is None:
                print("ASSISTANT: A selected item is unavailable.")
                return

            price = Decimal(str(item["price"]))
            quantity = selection["quantity"]

            new_items.append({
                "menuItemId": item["id"],
                "name": item["name"],
                "quantity": quantity,
                "unitPrice": str(price),
                "lineTotal": str(price * quantity)
            })

        session["items"].extend(new_items)
        session["pendingItems"].pop(0)

    # CASE 3: Confirm or cancel
    elif session["status"] == "awaiting_confirmation":
        import re

        normalized = re.sub(
            r"[^\w\s]", "", message.lower()
        )
        normalized = " ".join(normalized.split())

        if normalized in {
            "yes", "confirm", "confirm order",
            "yes confirm my order", "place order"
        }:
            session["status"] = "confirmed"
            print("ASSISTANT: Order confirmed in memory.")
            show_session(session_id)
            return

        if normalized in {"no", "cancel", "cancel order"}:
            session["status"] = "cancelled"
            print("ASSISTANT: Order draft cancelled.")
            return

        print("ASSISTANT: Please reply confirm or cancel.")
        return

    else:
        print("ASSISTANT: Session is already closed.")
        return

    # Determine next conversation step
    if session["pendingItems"]:
        issue = session["pendingItems"][0]
        session["status"] = "awaiting_clarification"

        if issue["status"] == "ambiguous":
            options = ", ".join(
                x["name"]
                for x in issue["options"]
                if x["available"]
            )
            print(f"ASSISTANT: Which would you like? {options}")
        else:
            print(
                "ASSISTANT: Cannot complete order. "
                f'{issue["requestedName"]} is '
                f'{issue["status"]}.'
            )
    elif session["items"]:
        session["status"] = "awaiting_confirmation"
        print(
            "ASSISTANT: Subtotal Rs.",
            calculate_subtotal(session),
            "- Confirm your order?"
        )
    else:
        print("ASSISTANT: No valid items were found.")
````
